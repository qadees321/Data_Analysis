## *List,Dictionary and Set Comprehension*

```python
"""Task: Use a for loop to print all numbers divisible by 7 from 1 to 100."""

num_list = [i for i in range(1,101) if (i % 7 == 0)]
print(num_list)
```

```python
"""Task: Calculate and print the sum of even numbers using a for loop from 1 to 50."""

sum_even = sum(i for i in range(1,51) if i % 2 == 0)
print(sum_even)
```

```python
"""Task: From 1 to 500, use a loop to find the first number divisible by both 13 and 7. Use break."""

first_number_divible_by_13_and_7 = next((i for i in range(1,500) if (i % 7 == 0) and (i % 13 == 0)), None)
print(first_number_divible_by_13_and_7)
```

```python
"""Task: From a list of words, create a dictionary where key is the word and value is the length, but
only for words longer than 4 letters."""

name_list = ["Steve", "Jean", "Katherine", "Paul","Huey", "Dewey", "Louie"]
name_dict = {i : len(i) for i in name_list if len(i) > 4}
print(name_dict)
```

```python
"""Task: Given a string, build a dictionary using comprehension where the keys are characters and
values are their frequency in the string."""

words = "Python is a versatile programming language"
words_dict = {i : i.count(i) for i in words.split()}

print(words_dict)
```

```python
"""Task: Create a list from 100 to 150 using list comprehension where prime numbers stay as-is, and
non-primes become "Not Prime"."""

prime_nonprime_numbers = [x if x > 1 and all(x % i != 0 for i in range(2, int(x**0.5) + 1)) else "Not Prime" for x in range(100,151)]
print(prime_nonprime_numbers)
```

```python
"""Task: Given a list l = list(range(1, 11)) and a check_even_odd(n) function already
created:
Create two new lists: one for even numbers and one for odd
numbers using the function.
Print both lists."""

def check_even_odd(num):
    even = []
    odd = []
    for n in num:
        if n % 2 == 0:
            even.append(n)
        else:
            odd.append(n)
    print(even)
    print(odd)
    
l = list(range(1, 11))
check_even_odd(l)
```

```python
"""Task: Write a function prime_upto(n) that returns all prime numbers up to n.
Include input validation for type and value."""

def prime_upto(n):
    prime_numbers = []
    for i in range(2,n):
        if all(i % j != 0 for j in range(2, int(i**0.5) + 1)):
            prime_numbers.append(i)
    return prime_numbers
print(prime_upto(100))
```

```python
"""Task: Write a function that accepts a list of numbers and returns a dictionary of:
total
average
max
min
all even numbers (in a list)"""

def analyze_numbers(num_list):
    total = sum(num_list)
    average = total / len(num_list)
    max_num = max(num_list)
    min_num = min(num_list)
    even_numbers = [num for num in num_list if num % 2 == 0]
    
    return {
        "total": total,
        "average": average,
        "max": max_num,
        "min": min_num,
        "even_numbers": even_numbers
    }

numbers = [10, 20, 30, 40, 50]
analyze_numbers(numbers)
```
