## *List Tasks* 

```python
"""Task: Grocery Stock Tracker: 
      available_stock = ["apple", "milk", "bread", "eggs", "rice", "butter"]
      Tasks:
          Add "flour" and "sugar" at the end. 
          Replace "bread" with "brown bread".
          Remove "milk" and print the updated list.
          Display only the middle 3 items using slicing."""

available_stock = ["apple","milk","bread","eggs","rice","butter"]
print(f"The available stock is: {available_stock}")
available_stock.extend(["flour","sugar"])
print(f"The available stock after adding flour and sugar is: {available_stock}")
print(f"The index of bread is: {available_stock.index("bread")}")
available_stock[2] = "brown bread"
print(f"After replacing bread with brown brown, the available stock is: {available_stock}")
available_stock.remove("milk")
print(f"After removing milk, the available stock is: {available_stock}")
print(f"The available_stock from index 2:5 is: {available_stock[2:5]}")
```

```python
"""Task: Grocery Stock Tracker: 
      ticket_prices = [250, 300, 200, 400, 150]
      Tasks:
           Find the highest and lowest ticket price.
           Sort the list in descending order.
           Add a new ticket price 350 while keeping the list sorted.
           Display only the top 3 expensive tickets."""

ticket_prices = [250, 300, 200, 400, 150]
print(f"The ticket prices are: {ticket_prices}")
print(f"The maximum ticket price is: {max(ticket_prices)}")
print(f"The minimum ticket price is: {min(ticket_prices)}")
ticket_prices.sort(reverse=True)
print(f"After sorting ticket prices in descending order: {ticket_prices}")
ticket_prices.append(350)
print(f"After appending '350' ticket price: {ticket_prices}")
ticket_prices.sort(reverse=True)
print(f"The ticket prices from index 0:3 are: {ticket_prices[0:3]}")
```

```python
"""Task: Playlist Management:
     You have to create a music playlist.
             Make a list of 5 songs.
             Swap the first and last song to change the playlist order.
             Add a new favorite song in the middle.
             Print all songs except the first and last."""

nfak_playlist = ["Nit Khair Manga","Mann Kunto Maula","Ali Mola Ali Dam Dam","Afreen Afreen","Allah Hoo Allah Hoo"]
print(f"The NFAK playlist is: {nfak_playlist}")
first = nfak_playlist[0]
last = nfak_playlist[-1]
nfak_playlist[-1] = first
nfak_playlist[0] = last
print(f"After swapping first song with last one, now the NFAK playlist is: {nfak_playlist}")
nfak_playlist[2] = "Sanu Ek Pal Chain Na Aave"
print(f"After adding a new song in the middle, now the NFAK playlist is: {nfak_playlist}")
print(f"The NFAK playlist from index 1:-1 is: {nfak_playlist[1:-1]}")
```

```python
"""Task: Product Prices: 
      You have prices of 6 products in your shop.
            Create a list of product prices.
            Increase the price of two products due to tax (1% tax /product).
            Remove the lowest-priced product.
            Calculate the total price of the remaining products."""

shop_products = ["Laptop", "Smartphone", "Headphones", "Smartwatch", "Keyboard", "Mouse"]
print(f"The shop products are: {shop_products}")
shop_prices = [150.00, 250.50, 99.99, 1200.00, 45.00, 550.25]
print(f"The shop prices are: {shop_prices}")
shop_prices[0] += (shop_prices[0]*0.01)
shop_prices[1] += (shop_prices[1]*0.01)
print(f"After increasing the price of two products by 1%, now the shop prices are: {shop_prices}")
shop_prices.sort()
shop_prices.pop(0)
shop_products.sort()
shop_products.pop(0)
print(f"After removing the lowest priced product, now the shop products are: {shop_products}")
print(f"After removing the lowest priced product, now the shop prices are: {shop_prices}")
print(f"The sum of the shop prices are: ${sum(shop_prices)}")
```

## *Sets Tasks*

```python
"""Task: A website wants to count how many unique visitors it had in a day.
      Every time a user visits, their user_id is logged — but users can visit multiple times.
      visits = [101, 102, 103, 101, 104, 102, 105, 103, 105, 106]
      Tasks
      - Find how many unique visitors visited the site.
      - Display the list of unique user_ids (in any order).
      - Compare the total visits vs. unique visitors."""

visits = [101, 102, 103, 101, 104, 102, 105, 103, 105, 106]
unique_visitors = set(visits)
len(unique_visitors)
unique_visitors
print(f'Total visitors: {visits}\nUnique visitors: {unique_visitors}')
print(f"Total visitors vs Unique visitors: {len(visits)} vs {len(unique_visitors)}")
```

```python
"""Task: Student Performance Tracker using Dictionaries:
Create a dictionary where:
Keys are student names.
Values are lists of their marks in 3 subjects.
Write a code to:
Add three new students with their marks.
Calculate average marks for each student.
Then add student(s) with the average marks into the dictionary.
Convert the dictionary into a list of tuples and sort it by average marks."""

student_marks = {
    "Zaryab" : [60,85,92],
    "Iqbal" : [49,57,88],
    "Asif" : [97,89,98]
}
print(f"Students with their marks are: \n{student_marks}")
average_marks_zaryab = round(sum(student_marks["Zaryab"])/len(student_marks["Zaryab"]), 2)
average_marks_iqbal = round(sum(student_marks["Iqbal"])/len(student_marks["Iqbal"]), 2)
average_marks_asif = round(sum(student_marks["Asif"])/len(student_marks["Asif"]), 2)
student_marks["Average_marks_zaryab"] = average_marks_zaryab
student_marks["Average_marks_iqbal"] = average_marks_iqbal
student_marks["Average_marks_asif"] = average_marks_asif
print(f"\nAfter adding every student's average mark in the dictionary: \n{student_marks}")
std_marks = list(tuple(student_marks.items()))
print(f"\nAfter converting the dictionary into list of tuples and sorting it by average marks: \n{sorted(std_marks[-3:], reverse=True),std_marks[0:3]}")

##print(sorted(std_marks))
```

```python
"""Task: Student Report Generator:
Create a system where users can:
Add student info (name, age, subjects, marks) using input function.
Store info in a dictionary.
Use sets to ensure unique subject names.
Use lists to maintain ordered scores.
Use tuples for immutable personal info (name, age).
Print a summary report at the end."""

student_report = {}
student_report["name"] = list(tuple(str(input("Please enter your name: ")).upper().split()))
student_report["age"] = list(tuple(input("Please enter your age: ").split()))
student_report["subjects"] = list(set(str(input("Please enter the names of the subjects: ")).upper().split()))
student_report["marks"] = list(input("Please enter the marks of the subjects: ").split())

print(f"Student Report: \n{student_report}")
```

```python
"""Task: Shopping Cart Snapshot:
Instructions:
Create 3 products.
Each product is a dictionary with:
"info" →
 a tuple: (name, category)
"price" →
 integer
"ratings" →
 list of 3 numbers
"tags" →
 set of 3 tags
Store all products in a list called cart.

Questions:
1.What is the category of the second product?
2.What is the total price of all products?
3.Find the average rating of the first product.
4.Combine all tags from the products (using |
operator).
5.Add a new product product4 with your own
data and update cart."""

product1 = {
        "info" : ("Keyboard", "Electronics"),
        "price" : 450,
        "ratings" : [5, 6,2],
        "tags" : {"Affordable", "Sustainable", "Excellent Quality"}
    }
product2 = {
        "info" : ("Rice", "Cooking"),
        "price" : 170,
        "ratings" : [3, 9,8],
        "tags" : {"Affordable", "Best Aroma", "Excellent Quality"}
        }
product3 = {
        "info" : ("Chair", "Furniture"),
        "price" : 1500,
        "ratings" : [7, 10,6],
        "tags" : {"Cheap", "Classical", "Wood Art"}
        }
cart = [product1,product2,product3]
cart
##1st question
print(f"The category of the second is: {cart[1]["info"][1]}")
##2nd question
total_price_of_all_products = cart[0]["price"]+cart[1]["price"]+cart[2]["price"]
print(f'Total price of all products: ${total_price_of_all_products}')
##3rd question
average_rating_of_first_product = round(sum(cart[0]["ratings"])/len(cart[0]["ratings"]),1)
print(f'Average rating of 1st product: {average_rating_of_first_product}')
##4th question
all_products_tags = cart[0]["tags"] | cart[1]["tags"] | cart[2]["tags"]
print(f'Tags of all products: {all_products_tags}')
##5th question
product4 = {
        "info" : ("Pen", "Office Supplies"),
        "price" : 50,
        "ratings" : [8, 9,8],
        "tags" : {"Beautiful", "Top Notch Writing", "Best for students"}
        }
cart.append(product4)
print("\nAfter adding fourth product, now the cart is:")
cart
```
