## *Conditional Statements and For & While loop Tasks*

```python
"""Task: Use a for loop to print all numbers divisible by 7 from 1 to 100."""
numbers = []
for i in range(1,101):
    if i % 7 == 0:
        numbers.append(i) 
print(f'These numbers are divisible by 7: {numbers}')
```

```python
"""Task: Calculate and print the sum of even numbers using a for loop from 1 to 50."""

sum = 0
for i in range(1,51):
    if i % 2 == 0:
         sum +=i
print(f"The sum of even numbers from 1 to 50 is: {sum}")
```

```python
"""Task: Ask the user for a password in a while loop. Allow 3 tries, and if the correct password is
entered, print "Access granted", otherwise print "Blocked"."""

correct_password = "Data_Analyst981()"
entered_password = input("Please enter your password: ")
tries = 3

while entered_password != correct_password and tries != 1:
    print("Incorrect Password! Please try again!")
    entered_password = input("Please enter your password: ")
    tries -= 1
    if tries == 1:
        print("Your account is blocked!")
    elif entered_password == correct_password:
        print("Access Granted!")
```

```python
"""Task: Create a countdown from 10 to 1 using a while loop. Use break if it reaches 5 and print
"Midpoint reached"."""

countdown = 10
while countdown > 0:
    countdown -= 1
    if countdown == 5:
        break

print(f"Midpoint reached: {countdown}")
```

```python
"""Task: From 1 to 500, use a loop to find the first number divisible by both 13 and 7. Use break."""

for i in range(1,501):
    if (i % 7 == 0) and (i % 13 == 0):
        print(f'The first number divisible by both 7 and 13 is: {i}')
        break
```

```python
"""Task: From a list of words, create a dictionary where key is the word and value is the length, but
only for words longer than 4 letters."""

name_list = ["Steve", "Jean", "Katherine", "Paul","Huey", "Dewey", "Louie"]
name_dict = {}
for char in name_list:
    if len(char) > 4:
        name_dict[char] = len(char)
print(name_dict)
```

```python
"""Task: Given a string, build a dictionary using comprehension where the keys are characters and
values are their frequency in the string."""
                                                                          
words = "Python is a versatile programming language"
words_dict = {}
for i in words.split():
    words_dict[i] = i.count(i)

print(words_dict)
```

```python
"""Task: Student Grade System
Create a program that:
Asks for student names and scores in a loop.
If score is below 0 or above 100, skip using continue.
After entering "done" as name, stop input using break.
Calculate and print:
Average score
Highest and lowest scores
Total number of students"""

student_name = []
scores = []

while True:
    name = input("Please enter the name of the student (or write 'done' when finished): ")
    if name.lower() == "done":
        break
    score = int(input(f"Please enter the score of {name}: "))
    if score < 0 or score > 100:    
        continue
    student_name.append(name)
    scores.append(score)

avg_score = sum(scores) / len(scores)
max_score = max(scores)
min_score = min(scores)
total_number_of_students = len(student_name)

print("----------------- Student Grade System -------------------")
print()
print(f"Average Score: {avg_score}")
print(f"Maximum Score: {max_score}")
print(f"Minimum Score: {min_score}")
print(f"Total number of students: {total_number_of_students}")
print(f"Students Names: {', '.join(student_name)}")
```

```python
"""Task: List Categorizer
Given a list of integers:
l = list(range(1,21))
Separate them into 3 lists:
Even
Odd
Multiples of 5
Print all 3 lists at the end."""

l = list(range(1,21))

even = [i for i in l if i % 2 == 0 ]
odd = [i for i in l if i % 2 != 0 ]
multiples_of_5 = [i for i in l if i % 5 == 0 ]

print(f"Even numbers list: {even}")
print(f"Odd numbers list: {odd}")
print(f"Multiples of 5 numbers list: {multiples_of_5}")
```

```python
"""Task: FizzBuzz Dictionatry
Make a dictionary for numbers 1 to 30 with:
"Fizz" if divisible by 3
"Buzz" if divisible by 5
"FizzBuzz" if divisible by both
"None" otherwise."""

numbers = list(range(1,31))
FizzBuzz_dictionary = {}
for i in numbers:
    if i % 3 == 0 and i % 5 == 0:
        FizzBuzz_dictionary[i] = "FizzBuzz"
    elif i % 5 == 0:
        FizzBuzz_dictionary[i] = "Buzz"
    elif i % 3 == 0:
        FizzBuzz_dictionary[i] = "Fizz"
    else:
        FizzBuzz_dictionary[i] = "None"

print(FizzBuzz_dictionary)
```
