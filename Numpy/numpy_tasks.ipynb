An online store wants to analyze its transaction data.

Dataset:
import numpy as np

    [1001, 501, 101,  2, 1500,  5, 200, 0],
    [1002, 502, 102,  1, 3200,  0, 150, 0],

    [1009, 506, 103,  7, 450, 10, 250, 0],

], dtype=float)
Your job
Calculate:
1.	Gross revenue for each transaction.

```python
import pandas as pd
```

```python
sales = np.array([
    [1001, 501, 101,  2, 1500,  5, 200, 0],
    [1002, 502, 102,  1, 3200,  0, 150, 0],
    [1003, 501, 103,  5,  450, 10, 250, 1],
    [1004, 503, 101,  3, 1500, 15, 200, 0],
    [1005, 504, 104,  2, 7800, 20, 300, 0],
    [1006, 505, 102,  4, 3200,  5, 150, 1],
    [1007, 502, 105, 10, 250,  0, 100, 0],
    [1008, 501, 104,  1, 7800, 25, 300, 0],
    [1009, 506, 103,  7, 450, 10, 250, 0],
    [1010, 503, 105,  8, 250, 5, 100, 0],
])

df_sales = pd.DataFrame(sales, columns=['transaction_id', 'customer_id', 'product_id','quantity', 'unit_price', 'discount_percent','shipping_cost', 'returned'])

#1.	Gross revenue for each transaction.
print("Gross Revenue for each transaction: ")
print(df_sales["quantity"]*df_sales["unit_price"])

#2.	Discount amount.
print("\nDiscount amount for each transaction: ")
print((df_sales["quantity"]*df_sales["unit_price"])*(df_sales["discount_percent"]/100))

#3.	Revenue after discount.
print("\nRevenue after discount for each transaction: ")
print(df_sales["quantity"]*df_sales["unit_price"]-((df_sales["quantity"]*df_sales["unit_price"])*(df_sales["discount_percent"]/100)))

#4.	Revenue after shipping.
print("\nRevenue after shipping cost for each transaction: ")
print(df_sales["quantity"]*df_sales["unit_price"]-df_sales["shipping_cost"])

#5.	Revenue lost from returned orders.
print(f"\nRevenue Lost from returned orders: ${np.sum(df_sales["quantity"]*df_sales["unit_price"]*df_sales["returned"])}")

#6.	Total company revenue.
print(f"\nTotal company revenue: ${np.sum(df_sales["quantity"]*df_sales["unit_price"])}")

#7.	Average transaction value.
print(f"\nAverage transaction value: ${np.mean(df_sales["quantity"]*df_sales["unit_price"])}")

#8.	Highest-value transaction.
print(f"\nHighest-value transaction: ${np.max(df_sales["quantity"]*df_sales["unit_price"])}")

#9.	Total spending of every customer.
gross_revenue = df_sales["unit_price"]*df_sales["quantity"]
unique_customers = np.unique(df_sales["customer_id"])
print("\nTotal spending of every customer:")
for customer in unique_customers:
    total = gross_revenue[df_sales["customer_id"] == customer].sum()
    print(f"Customer '{customer}' -> Total spending ${total}")

#10. Top 3 customers by spending.
spending = []
print("\nTotal spending of every customer:")
for customer in unique_customers:
    total = gross_revenue[df_sales["customer_id"] == customer].sum()
    spending.append(total)
spending = np.array(spending)
top_3_spending = np.argsort(spending)[::-1][:3]
print(f"Customers: {unique_customers[top_3_spending]}")
print(f"Spending: {spending[top_3_spending]}")

#11. Revenue generated by every product.
unique_products = np.unique(df_sales["product_id"])
print("\nRevenue generated by each product:")
for product in unique_products:
    total = gross_revenue[df_sales["product_id"] == product].sum()
    print(f"Product '{product}' -> Revenue generated ${total}")

#12. Return rate for every product.
print("\nReturn rate for each product:")
for product in unique_products:
    product_mask = df_sales["product_id"] == product
    total_transaction = np.sum(product_mask)
    returned_transaction = df_sales.loc[df_sales["product_id"] == product, "returned"].sum()
    return_rate = returned_transaction / total_transaction * 100

    print(f"Product '{product}' -> Return rate: {return_rate}%")

#13. Percentage contribution of each transaction to total revenue.
print("\nReturn rate for each product:")
transaction_id = df_sales["transaction_id"].to_numpy()
for t_id in transaction_id:
    t_mask = df_sales["transaction_id"] == t_id
    total_revenue = np.sum(df_sales["quantity"]*df_sales["unit_price"])
    t_id_revenue = df_sales.loc[df_sales["transaction_id"] == t_id, "unit_price"] * df_sales.loc[df_sales["transaction_id"] == t_id, "quantity"]

    transaction_contribution = round(t_id_revenue / total_revenue * 100, 2)
    print(f"transaction ID '{t_id}' -> Transaction contribution in revenue: {transaction_contribution.item()}%")
```
