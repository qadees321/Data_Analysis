```python
"""Task: Gym Membership Fee Calculator
      Ask the user:
          Enter registration fee (integer).
          Enter monthly fee (integer).
          Enter number of months they want to pay for.
    Calculate:
          Total cost = registration fee plus (monthly fee × number of months).
    Then:
         Ask if the user has a discount coupon (yes/no).
         If yes, reduce the total by 10% (use assignment operators).
         Print the final amount to be paid."""

registration_fee= int(input("Please enter your registration fee: "))
monthly_fee = int(input("Please enter your monthly fee: "))
number_of_months = int(input("Please enter the number of months you want to pay: "))

total_cost = registration_fee + (monthly_fee * number_of_months)
discount_coupon = input("Do you have any discount coupon? (yes/no): ").lower()
if discount_coupon == "yes":
    total_cost -= (total_cost*0.10)

print(f'Your final amount to be paid is: ${total_cost}')
```

```python
"""Task: Simple E-Wallet Recharge and Spending:
      Start with an e-wallet balance of 0.
      Ask the user:
               "How much money do you want to add?"
               "How much money did you spend on food?"
               "How much money did you spend on transport?"
               "How much money did you spend on shopping?"
      Use assignment operators (+=, -=) to update the wallet after each step.
     Then:
               Calculate whether the balance is more than 500 and less than 5000 (use logical   
               operator to print True or False).
               Print the final wallet balance."""

e_wallet_balance = 0
money_add = int(input("How much money do you want to add?"))
food_expenditure = int(input("How much money did you spend on food?"))
transort_expenditure = int(input("How much money did you spend on transport?"))
shopping_expenditure = int(input("How much money did you spend on shopping?"))
e_wallet_balance += money_add
e_wallet_balance -= food_expenditure
e_wallet_balance -= transort_expenditure
e_wallet_balance -= shopping_expenditure

if e_wallet_balance > 500 and e_wallet_balance < 5000:
    print(f'Your e-wallet balance is more than 500 and less than 5000: E-Wallet Balance: ${e_wallet_balance}')
elif e_wallet_balance > 5000:
    print(f'Your e-wallet balance is more than 5000: E-Wallet Balance: ${e_wallet_balance}')
else:
    print(f'E-Wallet Balance: ${e_wallet_balance}')
```

## *Modules*

```python
"""Task: Expiry Date Tracker
     Step 1: Ask the user:
                      manufacture date (YYYY-MM-DD)  and validity in days
      
     Step 2: Calculate:
                      The expiry date.
                      Whether the product is expired or valid today.
     Step 3: Display:
                      Print results in a formatted way using strftime()."""

from datetime import datetime, timedelta

manufacture_date = input("Please enter the manufacture date(In 'YYYY-MM-DD' format): ")
total_life_in_days = int(input("Please enter the life of the product in days: "))
manufacture_date = datetime.strptime(manufacture_date,"%Y-%m-%d")
expiry_date = manufacture_date + timedelta(days=total_life_in_days)
today = datetime.now()

if today > expiry_date:
    status = "This product is Expired!"
else:
    status = "This product is still valid!"

print("\n------------- Expiry Date Tracker ---------------")
print(f"Manufacture Date:     {manufacture_date.strftime('%Y-%B-%d')}")
print(f"Expiry Date:          {expiry_date.strftime('%Y-%B-%d')}")
print(f"Today's Date:         {today.strftime('%Y-%B-%d')}")
print(f"Status Today:         {status}")
```

```python
"""Task: Code Performance Timer
     Step 1: Ask the user:
                  Enter a number n.
     Step 2: Calculate:
                   Sum of squares of numbers up to n using following formula:
     
      Step 3: Display:
                    The computed sum of squares.
                    The execution time in both seconds and milliseconds."""
import time
start_time = time.perf_counter()

user_input = int(input("Please enter a number: "))
nth_number_of_squares = int(user_input*(user_input+1)*(2*user_input+1)/6)

end_time = time.perf_counter()
time_in_seconds = end_time - start_time
time_in_milliseconds = time_in_seconds * 1000

print(f"The computed sum of squares: {nth_number_of_squares}")
print(f"Execution time taken in seconds: {time_in_seconds:.3f} seconds")
print(f"Execution time taken in Milliseconds: {time_in_milliseconds:.3f} milliseconds")
```

```python
"""Task: OTP Verification System
Step 1: Ask the user:
Enter the Amount to send
Step 2: Generate:
Generate a 6-digit random OTP using random.randint().
Display it (simulate sending it to the user).
Start a timer of 10 seconds if the user doesn’t enter OTP, it expires.
Step 3: Display:
If OTP entered correctly and within time → print “✅ Amount Sent”
.
Otherwise → print “❌ OTP Expired or Incorrect” """

import random
import time
import threading

amount = input("Please enter the amount to send: ")
otp_code = random.randint(100000, 999999)
print(f"Your verification code(OTP) is: {otp_code}")
print("You have only 10 seconds to enter the OTP, otherwise your transaction will be rejected!")

entered_otp = [None]
def get_otp():
    entered_otp[0] = input("Please enter the OTP code: ")

thread = threading.Thread(target=get_otp)
thread.start()

thread.join(timeout=10)

if thread.is_alive():
    print("\nOTP Expired or Incorrect")
else:
    if entered_otp[0] == str(otp_code):
        print("\n----------------------------------- Amount Sent! --------------------------------------- ")
        print(f"------------------ You have successfully sent the amount worth ${amount} ------------------- ")
    else:
        print("\nOTP expired or incorrect!")
    
```
